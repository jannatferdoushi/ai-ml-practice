```python
from sklearn.datasets import load_diabetes  # It imports the Diabetes dataset from Scikit-learn.
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet  # This imports four regression algorithms.
from sklearn.model_selection import train_test_split # This imports the function used to divide the dataset.
from sklearn.metrics import r2_score # How well the model predicts the output.
```

```python
X,y = load_diabetes(return_X_y=True) # This loads the diabetes dataset.
# load_diabetes() loads the diabetes dataset from Scikit-learn.
# return_X_y=True tells Scikit-learn:"Return only the input features (X) and the target values (y)."
# X stores the input features (independent variables).
# y stores the target/output values (dependent variable).
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2) # This divides the dataset into two parts.
```

```python
# Linear Regression
reg = LinearRegression()  # This creates a Linear Regression object. 
reg.fit(X_train,y_train) # This is the training step.
y_pred = reg.predict(X_test) # Now the trained model predicts the outputs for the test data.
r2_score(y_test,y_pred) # This compares - Actual values and Predicted values 
```

```python
# Ridge 
reg = Ridge(alpha=0.1)  # Creates a Ridge Regression model. 
reg.fit(X_train,y_train)  # Learns coefficients using the Ridge loss function.
y_pred = reg.predict(X_test)  # Predicts outputs using the Ridge model.
r2_score(y_test,y_pred) # Calculates the R² score of Ridge Regression.
```

```python
# Lasso
reg = Lasso(alpha=0.01) # Creates a Lasso Regression model. 
reg.fit(X_train,y_train) # Learns coefficients while minimizing 
y_pred = reg.predict(X_test) # Predicts outputs.
r2_score(y_test,y_pred) # Calculates the R² score. 
```

```python
# ElasticNet
# alpha - Controls the overall regularization strength.
reg = ElasticNet(alpha=0.005,l1_ratio=0.9)  # Creates an ElasticNet model.
# l1_ratio-This controls how much Lasso and Ridge are mixed.
# l1_ratio = 1 → Pure Lasso
# l1_ratio = 0 → Pure Ridge
# l1_ratio = 0.5 → Equal mix of Lasso and Ridge
# l1_ratio = 0.9 → 90% Lasso effect and 10% Ridge effect

# So your model behaves mostly like Lasso but still includes a small Ridge component.
reg.fit(X_train,y_train) # Learns coefficients using the ElasticNet loss function:
y_pred = reg.predict(X_test) # Predicts the target values for the test data.
r2_score(y_test,y_pred) # Calculates the R² score of the ElasticNet model.
```
