```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score   # r2_score = A function that calculates the R² (R-squared) score of a regression model.
from sklearn.metrics import mean_squared_error
```

```python
from sklearn.datasets import load_diabetes

data=load_diabetes()
```

```python
print(data.DESCR)  # DESCR stands for Description.It contains a long text that explains everything about the dataset.
```

```python
X=data.data   # The first data is the variable that stores the entire diabetes dataset.
     # The second .data is an attribute of that dataset containing the input features.# "Give me the input features stored inside the diabetes dataset."

y=data.target # It means ->  The second .data is an attribute of that dataset containing the input features.# "Give me the input features stored inside the diabetes dataset."
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=45)
```

```python
from sklearn.linear_model import LinearRegression
L=LinearRegression()
```

```python
L.fit(X_train,y_train)  # L is an empty Linear Regression model.fit() -> Train the model using the training data.
```

```python
print(L.coef_)
print(L.intercept_)
```

```python
y_pred=L.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_squared_error    # r2_score → Measures how well the regression model fits the data.mean_squared_error → Measures how large the prediction errors are.

print("R2 score",r2_score(y_test,y_pred)) # y_test → Actual (correct) values y_pred → Predicted values from the model and returns a number called the R² (R-squared) score.r2_score measures how well a regression model's predictions match the actual values.
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred))) # This calculates the Mean Squared Error (MSE).
```

```python
from sklearn.linear_model import Ridge
R=Ridge(alpha=100000)
```

```python
R.fit(X_train,y_train)
```

```python
print(R.coef_)
print(R.intercept_)
```

```python
y_pred1=R.predict(X_test)
```

```python
print("R2 score",r2_score(y_test,y_pred1))
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred1)))
```

```python
m = 100
x1 = 5 * np.random.rand(m, 1) - 2
x2 = 0.7 * x1 ** 2 - 2 * x1 + 3 + np.random.randn(m, 1)

plt.scatter(x1, x2)
plt.show()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures

def get_preds_ridge(x1, x2, alpha):
    model = Pipeline([
        ('poly_feats', PolynomialFeatures(degree=16)),
        ('ridge', Ridge(alpha=alpha))
    ])
    model.fit(x1, x2)
    return model.predict(x1)

alphas = [0, 20, 200]
cs = ['r', 'g', 'b']

plt.figure(figsize=(10, 6))
plt.plot(x1, x2, 'b+', label='Datapoints')

for alpha, c in zip(alphas, cs):
    preds = get_preds_ridge(x1, x2, alpha)
    # Plot
    plt.plot(sorted(x1[:, 0]), preds[np.argsort(x1[:, 0])], c, label='Alpha: {}'.format(alpha))

plt.legend()
plt.show()
```
